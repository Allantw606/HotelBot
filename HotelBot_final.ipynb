
# 🏨 Hotel Assistant Bot

## Chatbot especialista em atendimento hoteleiro utilizando Google Gemini

### Objetivo do projeto

Desenvolver um chatbot capaz de responder perguntas de clientes utilizando uma base de conhecimento privada.

O chatbot foi configurado para:

- Utilizar somente informações fornecidas no contexto;
- Evitar respostas inventadas (alucinações);
- Responder dúvidas sobre hospedagem;
- Encerrar a conversa após três perguntas com um resumo final.

### Tecnologia utilizada

- Google Gemini API
- Python
- Google Colab

```python
from google import genai

# Insira sua chave da API do Gemini aqui antes de executar
GOOGLE_API_KEY = "SUA_CHAVE_API_AQUI"

client = genai.Client(
    api_key=GOOGLE_API_KEY
)

print("✅ Gemini conectado com sucesso!")
```

# 📚 Base de conhecimento privada

O chatbot possui acesso somente às informações do Hotel Vista Azul.

As informações abaixo representam o conhecimento fornecido ao modelo.

```python
conhecimento = """

Você é o Hotel Assistant, atendente virtual do Hotel Vista Azul.

PERSONALIDADE:
Você é educado, profissional e prestativo.

OBJETIVO:
Ajudar clientes respondendo dúvidas sobre o Hotel Vista Azul.

Utilize somente as informações abaixo.

=========================
INFORMAÇÕES DO HOTEL
=========================

Nome:
Hotel Vista Azul

Localização:
Cidade fictícia Porto Azul.

Recepção:
Funcionamento 24 horas.


=========================
ACOMODAÇÕES
=========================

Quarto Standard:

- Capacidade: até 2 pessoas.
- Cama de casal.
- Televisão.
- Ar-condicionado.
- Banheiro privativo.

Valor:
R$250,00 por diária.


Quarto Luxo:

- Capacidade: até 3 pessoas.
- Cama king size.
- Frigobar.
- Ar-condicionado.
- Varanda.

Valor:
R$400,00 por diária.


Suíte Premium:

- Capacidade: até 4 pessoas.
- Sala de estar.
- Varanda com vista para o mar.
- Frigobar.
- Serviço de quarto.

Valor:
R$650,00 por diária.


=========================
REGRAS DE HOSPEDAGEM
=========================

Check-in:
A partir das 14:00.

Check-out:
Até 12:00.

Cancelamento:

Até 48 horas antes da hospedagem não possui cobrança.


=========================
SERVIÇOS
=========================

- Café da manhã incluso.
- Wi-Fi gratuito.
- Estacionamento gratuito.
- Serviço de quarto das 07:00 às 22:00.
- Piscina das 08:00 às 20:00.


=========================
PAGAMENTO
=========================

Aceita:

- Pix.
- Cartão de crédito.
- Cartão de débito.


=========================
REGRAS DO CHATBOT
=========================

Nunca invente informações.

Caso a resposta não esteja neste documento, responda:

"Essa informação não está disponível na minha base de conhecimento."

O cliente fará apenas três perguntas.

Após a terceira pergunta:
Faça um resumo das respostas e encerre a conversa.

"""
```

# 🤖 Inicialização do Hotel Assistant

Nesta etapa o modelo Gemini recebe:
- Personalidade;
- Objetivo;
- Regras;
- Conhecimento privado.

```python
historico = []

quantidade_perguntas = 0

print("🏨 Hotel Assistant iniciado!")
print("Faça até 3 perguntas sobre o hotel.")
```

```python
while quantidade_perguntas < 3:

    pergunta = input("Cliente: ")

    prompt = f"""
    {conhecimento}

    Histórico da conversa:
    {historico}

    Pergunta do cliente:
    {pergunta}

    Responda seguindo todas as regras.
    """

    resposta = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    print("\n🤖 Hotel Assistant:")
    print(resposta.text)

    historico.append({
        "Cliente": pergunta,
        "Bot": resposta.text
    })

    quantidade_perguntas += 1

print("\n✅ Limite de perguntas atingido. Gerando resumo final...")
```

# 📋 Resumo da conversa

Após três perguntas, o chatbot gera um resumo das informações apresentadas.

```python
resumo_prompt = f"""

Com base na conversa abaixo:

{historico}


Faça um resumo das três respostas fornecidas.

Depois informe:

"Conversa encerrada."

"""

resumo = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=resumo_prompt
)


print("\nResumo final:")
print(resumo.text)
```
